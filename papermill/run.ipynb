```python
# !pip install papermill
import papermill as pm

for a, b in zip(range(3), range(3)):
    pm.execute_notebook(
        "example.ipynb",
        f"example_{a}_{b}.ipynb",
        parameters={"a": a, "b": b},
    )

    data = pm.inspect_notebook(f"example_{a}_{b}.ipynb")
    print(data)
  
```
