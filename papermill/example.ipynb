# Example notebook

- Code cell with parameters must be tagged with tag "parameters"
- Papermill finds the "parameters" cell and adds a new cell tagged with "injection-parameters" after this cell
- The new cell has the altered parameters.

```python
import scrapbook as sb

# Parameter
a = 1
b = 2
```

## Task
- We do a task to create a new variable.

```python
# Results
c = a + b
print(c)
```

## Glue 
- the value of c is marked as "result"
- with scrapbook "result" can be read from another notebook later

```python
sb.glue("result", c)
```
