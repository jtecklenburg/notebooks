# Papermill & Scrapbook example

- overwrites variables in notebook
- runs altered notebook
- collect results from executed notebook

```python
# !pip install papermill
import papermill as pm
import scrapbook as sb

for a, b in zip([3, 5, 6], [9, 5, 3]):
    # overwrite the notebook with the new parameters and run it    
    pm.execute_notebook(
        "example.ipynb",
        f"example_{a}_{b}.ipynb",
        parameters={"a": a, "b": b},
    )

    # collect the results from the executed notebook
    data = sb.read_notebook(f"example_{a}_{b}.ipynb")
    print(data.scraps.data_dict["result"])  
```
